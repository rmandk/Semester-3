# 1. Variable

Variable adalah nama atau alias yang kita sematkan pada lokasi memori yang menyimpan nilai/data. Proses pembuatan variable dilakukan melalui 2 tahapan, yakni:

- pemberian nama (identifier) pada variable
- penetapan nilai awal variable (initialization)

Kita dapat menetapkan nilai apapun pada sebuah variable. Namun, kita tidak dapat memberikan nama variable secara sembarangan. Berikut ini adalah *rules* yang harus kita ikuti dalam pemberian nama variable:

- tidak boleh diawali dengan angka
- tidak boleh mengggunakan special character, kecuali simbol *underscore* (`_`)
- tidak boleh menggunakan spasi
- tidak boleh menggunakan keyword Python

String pendukung dapat ditambahkan ketika kita mencetak nilai variable.

printvarprompt1.PNG

```python
# 5.2
```

```python
nama = "Beckham Putra"
usia = 23
bmi = 25.1
pemain_persib = True

print("Nama\t\t:", nama)
print("Usia\t\t:", usia, 'tahun')
print("BMI\t\t:", bmi)
print("Pemain Persib\t:", pemain_persib)
```

Selain menggunakan simbol koma (`,`) untuk menambahkan string dalam pencetakan nilai variable. Masih ada metode lain yang bisa kita terapkan, diantaranya:

- menggunakan *string formatting* dengan bantuan simbol `{}`


string formatting.PNG

```python
# 5.3
```

```python
nama = "Beckham Putra"
usia = 23
bmi = 25.1
pemain_persib = True

print("Nama\t\t: {}".format(nama))
print("Usia\t\t: {} tahun".format(usia))
print("BMI\t\t: {}".format(bmi))
print("Pemain Persib\t: {}".format(pemain_persib))
```

- menggunakan `f-string`

f-string.PNG

```python
# 5.4
```

```python
nama = "Beckham Putra"
usia = 23
bmi = 25.1
pemain_persib = True

print(f"Nama\t\t: {nama}")
print(f"Usia\t\t: {usia} tahun")
print(f"BMI\t\t: {bmi}")
print(f"Pemain Persib\t: {pemain_persib}")
```

# 7. Menginput Nilai

Sebelumnya, kita telah membahas cara memberikan/menetapkan nilai pada sebuah variable. Kita dapat memberikan/menetapkan nilai tersebut dengan cara menulisnya langsung dalam code kita. Metode ini dikenal dengan sebutan *hard coding*.

Sayangnya, hard coding hanya bisa dilakukan oleh kita, sebagai programmer. Karena, kita memiliki akses untuk mengedit program yang kita tulis.

User (orang lain yang hanya menggunakan aplikasi kita) tentunya tidak dapat memberikan/menetapkan nilai sebuah variable melalui metode hard coding. Salah satu metode yang memungkinkan mereka melakukan hal tersebut adalah dengan melalui perintah `input()`.


input_var.PNG

```python
# 7.1
```

```python
nama = input("Nama\t\t: ")
usia = input("Usia\t\t: ")
jenkel = input("Jenis Kelamin\t: ")
asal_pt = input("Asal Perguruan Tinggi: ")
prodi = input("Program Studi\t: ")

print(f"""Halo! Perkenalkan, namaku {nama} dan usiaku {usia} tahun.
Aku seorang {jenkel}.
Saat ini, aku berkuliah di {asal_pt}, jurusan {prodi}.
Salam kenal!
""")
```
