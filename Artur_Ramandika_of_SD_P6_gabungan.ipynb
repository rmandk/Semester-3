```python
# Data mahasiswa dan kehadiran
mahasiswa_terdaftar = ["Andi", "Budi", "Citra", "Dewi", "Pandu"]
kehadiran = set()

def tampilkan_terdaftar():
    print("\nDaftar Mahasiswa Terdaftar:")
    for mhs in mahasiswa_terdaftar:
        print("-", mhs)

def tambah_kehadiran():
    nama = input("Masukkan nama mahasiswa yang hadir: ")
    if nama in mahasiswa_terdaftar:
        kehadiran.add(nama)
        print(f"{nama} telah ditandai hadir.")
    else:
        print("Mahasiswa tidak terdaftar!")

def tampilkan_kehadiran():
    print("\nDaftar Mahasiswa Hadir:")
    if kehadiran:
        for mhs in kehadiran:
            print("-", mhs)
    else:
        print("Belum ada yang hadir.")

def rekap_kehadiran():
    total = len(mahasiswa_terdaftar)
    hadir = len(kehadiran)
    print("\n=== Rekap Kehadiran ===")
    print(f"Total Mahasiswa: {total}")
    print(f"Jumlah Hadir   : {hadir}")
    print(f"Persentase     : {(hadir/total)*100:.2f}%")

# Menu utama
def menu():
    while True:
        print("\n ********************************************")
        print(" ** Program Manajemen Kehadiran Mahasiswa **")
        print(" ** Menggunakan List, Tuple, Set, Dictionary **")
        print(" ********************************************")

        print("\n ======= MENU KEHADIRAN MAHASISWA =====")
        print("1. Tampilkan Mahasiswa Terdaftar")
        print("2. Tambah Mahasiswa Hadir")
        print("3. Lihat Daftar Hadir")
        print("4. Rekap Kehadiran")
        print("5. Keluar")

        pilihan = input("Pilih menu (1-5): ")

        if pilihan == "1":
            tampilkan_terdaftar()
        elif pilihan == "2":
            tambah_kehadiran()
        elif pilihan == "3":
            tampilkan_kehadiran()
        elif pilihan == "4":
            rekap_kehadiran()
        elif pilihan == "5":
            print("Terima kasih! Program selesai.")
            break
        else:
            print("Pilihan tidak valid. Silakan coba lagi.")

# Jalankan program
menu()
```
